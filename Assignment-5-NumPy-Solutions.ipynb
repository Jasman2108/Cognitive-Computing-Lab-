## Q1. Create a 1D NumPy array and perform basic operations

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print('Original array:', arr)
print('Addition of 2:', arr + 2)
print('Multiplication by 3:', arr * 3)
print('Division by 2:', arr / 2)
```

## Q2. Basic NumPy array questions

```python
# (a) Reverse the array
arr = np.array([1, 2, 3, 6, 4, 5])
print('Original array:', arr)
print('Reversed array:', arr[::-1])

# (b-i) Most frequent value and its indices in x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
modes = values[counts == max_count]
print('\nx:', x)
for value in modes:
    print(f'Most frequent value in x: {value}; count: {np.count_nonzero(x == value)}; indices: {np.where(x == value)[0]}')

# (b-ii) Most frequent value and its indices in y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
modes = values[counts == max_count]
print('\ny:', y)
for value in modes:
    print(f'Most frequent value in y: {value}; count: {np.count_nonzero(y == value)}; indices: {np.where(y == value)[0]}')
```

## Q3. Access elements in a 2D array using row and column indices

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print('Array:\n', arr)
# Python indices start at 0: 1st row, 2nd column -> [0, 1]
print('1st row, 2nd column:', arr[0, 1])
# 3rd row, 1st column -> [2, 0]
print('3rd row, 1st column:', arr[2, 0])
```

## Q4. Create a named array using linspace() and inspect its properties

```python
# Python variable names cannot contain spaces, so use a valid variable name.
jasmandeep_sharma = np.linspace(10, 100, 25)
print('Array:', jasmandeep_sharma)
print('Dimensions:', jasmandeep_sharma.ndim)
print('Shape:', jasmandeep_sharma.shape)
print('Total elements:', jasmandeep_sharma.size)
print('Data type:', jasmandeep_sharma.dtype)
print('Total bytes consumed:', jasmandeep_sharma.nbytes)

# A 1D array has no different 2D transpose; reshape it to a column vector first.
transposed_by_reshape = jasmandeep_sharma.reshape(1, 25).T
print('\nTranspose using reshape() and T:')
print(transposed_by_reshape)

# The T attribute alone does not change the shape of a 1D array.
print('\nUsing .T directly on the 1D array:')
print(jasmandeep_sharma.T)
print('Shape of direct .T:', jasmandeep_sharma.T.shape)
print('Answer: .T alone does not transpose a 1D array; reshape first to make it 2D.')
```
