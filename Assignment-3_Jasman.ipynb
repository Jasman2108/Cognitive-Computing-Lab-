## Q6. Create employees.csv

```python
employees_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employees_data)
employees.to_csv('employees.csv', index=False)
employees
```

## Q6(a–d). Shape, summary, descriptive statistics, first and last rows

```python
# a) Number of rows and columns
print('Shape:', employees.shape)

# b) Data types and non-null counts
employees.info()

# c) Descriptive statistics
display(employees.describe(include='all'))

# d) First 5 rows and last 3 rows
print('First 5 rows:')
display(employees.head(5))
print('Last 3 rows:')
display(employees.tail(3))
```

## Q6(e). Calculate statistics

```python
print('Average salary:', employees['Salary'].mean())
print('Total bonus paid:', employees['Bonus'].sum())
print('Youngest employee age:', employees['Age'].min())
print('Highest performance rating:', employees['Rating'].max())
```

## Q6(f). Sort by salary in descending order

```python
employees = employees.sort_values(by='Salary', ascending=False)
employees
```

## Q6(g). Add a performance category

```python
def performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    return 'Average'

employees['Performance_Category'] = employees['Rating'].apply(performance_category)
employees
```

## Q6(h). Identify missing values

```python
employees.isnull().sum()
```

## Q6(i). Rename Employee_ID to ID

```python
employees = employees.rename(columns={'Employee_ID': 'ID'})
employees
```

## Q6(j). Filter employees

```python
# i. Employees with more than 5 years of experience
experienced_employees = employees[employees['Years_of_Experience'] > 5]
display(experienced_employees)

# ii. Employees in the IT department
it_employees = employees[employees['Department'] == 'IT']
display(it_employees)
```

## Q6(k). Add a Tax column deducting 10% of Salary

```python
# Tax is the amount deducted (10% of salary), as stated in the assignment.
employees['Tax'] = employees['Salary'] * 0.10
employees
```

## Q6(l). Save the modified DataFrame to a new CSV file

```python
employees.to_csv('employees_modified.csv', index=False)
print('Saved modified dataset as employees_modified.csv')
```
