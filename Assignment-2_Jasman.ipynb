## Question 1 — List operations

```python
# Create the list
L = [10, 20, 30, 40, 50, 60, 70, 80]

# i. Add 200 and 300
L.extend([200, 300])
print('After adding 200 and 300:', L)

# ii. Remove 10 and 30
L.remove(10)
L.remove(30)
print('After removing 10 and 30:', L)

# iii. Sort in ascending order
L.sort()
print('Ascending order:', L)

# iv. Sort in descending order
L.sort(reverse=True)
print('Descending order:', L)
```

## Question 2 — Tuple operations

```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

# i. Highest score and its index
highest = max(scores)
print('Highest score:', highest)
print('Index of highest score:', scores.index(highest))

# ii. Lowest score and how many times it appears
lowest = min(scores)
print('Lowest score:', lowest)
print('Count of lowest score:', scores.count(lowest))

# iii. Reverse the tuple and return it as a list
reversed_scores = list(reversed(scores))
print('Reversed tuple as a list:', reversed_scores)

# iv. Check whether a user-entered score is present
try:
    target = float(input('Enter a score to search for (e.g. 76): '))
    if target in scores:
        print('Score found at index:', scores.index(target))
    else:
        print('Score is not present in the tuple.')
except ValueError:
    print('Please enter a valid number.')
```

## Question 3 — Random numbers, odd/even and prime counts

```python
import random

# Create a list of 100 random integers between 100 and 900 (inclusive)
numbers = [random.randint(100, 900) for _ in range(100)]
print('Random numbers:', numbers)

# i. All odd numbers
odd_numbers = [n for n in numbers if n % 2 != 0]
print('\nOdd numbers:', odd_numbers)
print('Count of odd numbers:', len(odd_numbers))

# ii. All even numbers
even_numbers = [n for n in numbers if n % 2 == 0]
print('\nEven numbers:', even_numbers)
print('Count of even numbers:', len(even_numbers))

# iii. All prime numbers
def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, int(n ** 0.5) + 1):
        if n % divisor == 0:
            return False
    return True

prime_numbers = [n for n in numbers if is_prime(n)]
print('\nPrime numbers:', prime_numbers)
print('Count of prime numbers:', len(prime_numbers))
```

## Question 4 — Set operations

```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

# i. Union: unique scores achieved by either team
print('Union (A | B):', A.union(B))

# ii. Intersection: scores common to both teams
print('Intersection (A & B):', A.intersection(B))

# iii. Symmetric difference: scores exclusive to either team
print('Symmetric difference:', A.symmetric_difference(B))

# iv. Check subset and superset relationships
print('Is A a subset of B?', A.issubset(B))
print('Is B a superset of A?', B.issuperset(A))

# v. Remove a user-entered score from A if it exists
try:
    x = int(input('Enter a score to remove from set A: '))
    if x in A:
        A.remove(x)
        print('Updated set A:', A)
    else:
        print(f'{x} is not present in set A.')
except ValueError:
    print('Please enter a valid integer.')
```
