## Q1. Sensor readings and vectorization

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# (a) Add the average measurement error of +2°C using vectorization
corrected_temperature = temperature + 2
print('Corrected temperatures (°C):', corrected_temperature)

# (b) Convert corrected Celsius readings to Fahrenheit
fahrenheit = (corrected_temperature * 9 / 5) + 32
print('Temperatures (°F):', fahrenheit)

# (c) Identify readings greater than 32°C using Boolean indexing
readings_above_32 = corrected_temperature[corrected_temperature > 32]
print('Corrected readings above 32°C:', readings_above_32)

# (d) Count readings greater than 32°C
count_above_32 = np.count_nonzero(corrected_temperature > 32)
print('Count above 32°C:', count_above_32)
```

**(e) Explanation:** Vectorization applies an operation to an entire NumPy array at once, without writing an explicit Python loop. Boolean indexing uses a condition to select only values that satisfy it. These approaches make code shorter and are often faster for large arrays because NumPy performs operations in optimized compiled code.

## Q2. Daily steps recorded by four users

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# (a) Total steps recorded
print('Total steps:', np.sum(steps))

# (b) Mean number of steps
print('Mean steps:', np.mean(steps))

# (c) Maximum and minimum values
print('Maximum steps:', np.max(steps))
print('Minimum steps:', np.min(steps))

# (d) Total steps for each day (sum down the rows)
print('Total steps for each day:', np.sum(steps, axis=0))

# (e) Total steps for each user (sum across columns)
print('Total steps for each user:', np.sum(steps, axis=1))

# (f) Position of the maximum value
flat_index = np.argmax(steps)
user_index, day_index = np.unravel_index(flat_index, steps.shape)
print('Maximum value flat index:', flat_index)
print('User/day position (zero-based):', (user_index, day_index))
print('User number:', user_index + 1, '| Day number:', day_index + 1)
```

## Q3. NumPy slicing, copying, reshaping and flattening

```python
# (a) Create the original array
original = np.array([1, 2, 3, 4, 5, 6])

# (b) Slice indices 1 through 4 (stop index is exclusive)
subset = original[1:5]

# (c) Modify the first element of subset
subset[0] = 999
print('Original after modifying subset:', original)
print('Subset:', subset)
# A basic slice is a view, so the original array changes too.

# (d) Create a separate copy, then modify it
copied_array = original[1:5].copy()
copied_array[0] = 500
print('\nOriginal after modifying copied array:', original)
print('Copied array:', copied_array)

# (e) Create values 1 to 12 and reshape to 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)
print('\n3 x 4 matrix:\n', matrix)

# (f) Extract requested rows, columns and submatrix
print('First row:', matrix[0, :])
print('Last row:', matrix[-1, :])
print('Second column:', matrix[:, 1])
# Rows 1–2 and columns 2–3 are interpreted as one-based positions.
print('Rows 1–2 and columns 2–3:\n', matrix[0:2, 1:3])

# (g) Flatten with flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()
print('\nflatten():', flattened)
print('ravel():', raveled)

# (h) Modify ravel output and check original
raveled[0] = 1000
print('\nAfter changing ravel output:')
print('raveled:', raveled)
print('matrix:', matrix)  # ravel usually returns a view for this contiguous array

# (i) Modify flatten output and check original
flattened[1] = 2000
print('\nAfter changing flatten output:')
print('flattened:', flattened)
print('matrix:', matrix)  # flatten returns a copy; matrix is unaffected

# (j) Array properties
print('\nShape:', matrix.shape)
print('Dimensions (ndim):', matrix.ndim)
print('Size:', matrix.size)
print('Data type:', matrix.dtype)
```

## Q4. Assistance score prediction using Ordinary Least Squares

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# (a) Shape and dimensions
print('Shape of X:', X.shape)
print('Dimensions of X:', X.ndim)

# (b) Transpose: rows become columns and columns become rows
X_T = X.T
print('\nX transpose:\n', X_T)

# (c) Matrix product X.T @ X
XTX = X_T @ X
print('\nX.T @ X:\n', XTX)

# (d) Compute inverse when possible
print('\nDeterminant of X.T @ X:', np.linalg.det(XTX))
try:
    XTX_inverse = np.linalg.inv(XTX)
    print('Inverse of X.T @ X:\n', XTX_inverse)
except np.linalg.LinAlgError:
    XTX_inverse = None
    print('X.T @ X is singular, so its inverse does not exist.')

# (e) Ordinary Least Squares: beta = (X.T X)^(-1) X.T y
# If the inverse is unavailable, use the equivalent least-squares solution.
if XTX_inverse is not None:
    beta = XTX_inverse @ X_T @ y
else:
    beta = np.linalg.lstsq(X, y, rcond=None)[0]
print('\nOLS coefficients (without intercept):', beta)

# (f) Coefficients correspond to sleep hours, activity level, and stress level.
print('Coefficient for sleep hours:', beta[0])
print('Coefficient for activity level:', beta[1])
print('Coefficient for stress level:', beta[2])
print('Each coefficient estimates the change in predicted score for a one-unit increase in that feature, holding the others fixed.')

# (g) Predict for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print('Predicted assistance score:', predicted_score)
```
